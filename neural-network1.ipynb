#勾配を求める

```python
def numerical_gradient(f, x):
  """
  Parameters
  ----------
  f : function
      入力: numpy配列
      出力: numpy配列
  x : numpy.ndarray
  """
  h = 1e-4
  grad = np.zeros_like(x)

  for idx in range(x.size):
    tmp_val = x[idx]
    # f(x+h)の計算
    x[idx] = tmp_val + h
    fxh1 = f(x)

    # f(x-h)の計算
    x[idx] = tmp_val - h
    fxh2 = f(x)

    grad[idx] = (fxh1 - fxh2) / (2*h)
    x[idx] = tmp_val

  return grad
```

```python
def func(x):
  """
  2次元配列(numpy配列)
  """
  return x[0]**2 + x[1]**2

print("例1:", numerical_gradient(func, np.array([3.0, 4.0])))
print("例2:", numerical_gradient(func, np.array([0.0, 2.0])))
print("例3:", numerical_gradient(func, np.array([3.0, 0.0])))
```

#勾配降下法

```python
def gradient_descent(f, init_x, lr=0.01, step_num=100):
  """
  f: function
  init_x: xの初期値
  lr: 学習率
  step_num: 学習回数
  """
  x = init_x

  for i in range(step_num):
    grad = numerical_gradient(f, x)
    x -= lr * grad

  return x
```

##例）関数の極小値を求める

```python
init_x = np.array([-3.0, 4.0])
x = gradient_descent(func, init_x, lr=0.1, step_num=100)

print(x)
```

#ソフトマックス関数・クロスエントロピー関数

## ソフトマックス関数

$$
\mathrm{softmax}(x_i) = \frac{e^{x_i}}{\sum_{j=1}^{n} e^{x_j}}
$$

数値安定化のための実装では、以下を用いる：

$$
\mathrm{softmax}(x_i) = \frac{e^{x_i - \max(x)}}{\sum_{j=1}^{n} e^{x_j - \max(x)}}
$$



```python
import numpy as np

def softmax(x):
    """
    x: numpy.ndarray (n,) または (batch, n)
    return: softmax適用結果（同じshape）
    """
    # 数値安定化（オーバーフロー防止）
    x = x - np.max(x, axis=-1, keepdims=True)

    exp_x = np.exp(x)
    sum_exp = np.sum(exp_x, axis=-1, keepdims=True)

    return exp_x / sum_exp
```

## クロスエントロピー損失関数

分類問題におけるクロスエントロピー：

$$
L = - \sum_{i=1}^{n} y_i \log p_i
$$

ここで：

- $y_i$：正解ラベル（one-hotベクトル）
- $p_i$：予測確率（softmaxの出力）

---

バッチデータの場合：

$$
L = - \frac{1}{N} \sum_{k=1}^{N} \sum_{i=1}^{n} y_{k,i} \log p_{k,i}
$$

```python
import numpy as np

def cross_entropy(y_true, y_pred):
    """
    y_true: (n,) または (batch, n) one-hot
    y_pred: (n,) または (batch, n) softmax出力

    return: スカラー損失
    """
    # log(0)防止
    eps = 1e-15
    y_pred = np.clip(y_pred, eps, 1 - eps)

    return -np.sum(y_true * np.log(y_pred)) / y_true.shape[0]
```

## softmax + クロスエントロピーの勾配

softmax：
$$
p_i = \frac{e^{z_i}}{\sum_j e^{z_j}}
$$

クロスエントロピー：
$$
L = - \sum_i y_i \log p_i
$$

このとき、入力 $z$ に対する勾配は：

$$
\frac{\partial L}{\partial z_i} = p_i - y_i
$$

## softmax + クロスエントロピーの勾配導出

## 前提

softmax：
$$
p_i = \frac{e^{z_i}}{\sum_k e^{z_k}}
$$

クロスエントロピー：
$$
L = -\sum_i y_i \log p_i
$$

---

## ステップ①：logを展開

$$
\log p_i = \log \left( \frac{e^{z_i}}{\sum_k e^{z_k}} \right)
= z_i - \log \left( \sum_k e^{z_k} \right)
$$

これを損失関数に代入：

$$
L = -\sum_i y_i \left( z_i - \log \sum_k e^{z_k} \right)
$$

展開すると：

$$
L = -\sum_i y_i z_i + \sum_i y_i \log \sum_k e^{z_k}
$$

---

## ステップ②：one-hotの性質

one-hotベクトルでは：

$$
\sum_i y_i = 1
$$

したがって：

$$
\sum_i y_i \log \sum_k e^{z_k}
= \log \sum_k e^{z_k}
$$

よって：

$$
L = -\sum_i y_i z_i + \log \sum_k e^{z_k}
$$

---

## ステップ③：微分

### 第1項

$$
\frac{\partial}{\partial z_j} \left( -\sum_i y_i z_i \right)
= -y_j
$$

---

### 第2項

$$
\frac{\partial}{\partial z_j} \log \sum_k e^{z_k}
= \frac{1}{\sum_k e^{z_k}} \cdot e^{z_j}
$$

$$
= \frac{e^{z_j}}{\sum_k e^{z_k}} = p_j
$$

---

## ステップ④：まとめ

$$
\frac{\partial L}{\partial z_j}
= -y_j + p_j
$$

---

## 結論

$$
\frac{\partial L}{\partial z} = p - y
$$

---

## ポイント

- logを先に展開すると計算が簡単になる
- one-hotの性質で式が大きく簡約される
- 結果は非常にシンプルな形になる

---

## 直感的理解

- 勾配 = 予測 - 正解
- 予測が大きすぎる → 減らす
- 小さすぎる → 増やす

# 1層ニューラルネットの勾配導出（W, b）

## モデル

入力：
$$
x \in \mathbb{R}^d
$$

重み・バイアス：
$$
W \in \mathbb{R}^{d \times n}, \quad b \in \mathbb{R}^n
$$

線形変換：
$$
z = xW + b
$$

softmax：
$$
p = \mathrm{softmax}(z)
$$

損失関数（クロスエントロピー）：
$$
L = -\sum_i y_i \log p_i
$$

---

## 既知の結果（重要）

$$
\frac{\partial L}{\partial z} = p - y
$$

---

## ステップ①：bの勾配

$$
z = xW + b
$$

なので：

$$
\frac{\partial z}{\partial b} = I
$$

よって：

$$
\frac{\partial L}{\partial b}
= \frac{\partial L}{\partial z}
= p - y
$$

---

## ステップ②：Wの勾配

$$
z = xW + b
$$

各成分：
$$
z_j = \sum_i x_i W_{i,j} + b_j
$$

---

### 微分

$$
\frac{\partial z_j}{\partial W_{i,j}} = x_i
$$

---

### 連鎖律

$$
\frac{\partial L}{\partial W_{i,j}}
= \frac{\partial L}{\partial z_j} \cdot \frac{\partial z_j}{\partial W_{i,j}}
$$

$$
= (p_j - y_j) \cdot x_i
$$

---

## 行列形式

$$
\frac{\partial L}{\partial W} = x^T (p - y)
$$

---

## まとめ

$$
\frac{\partial L}{\partial z} = p - y
$$

$$
\frac{\partial L}{\partial b} = p - y
$$

$$
\frac{\partial L}{\partial W} = x^T (p - y)
$$

---

## バッチ版

入力：
$$
X \in \mathbb{R}^{N \times d}
$$

$$
Z = XW + b
$$

$$
P = \mathrm{softmax}(Z)
$$

---

勾配：

$$
\frac{\partial L}{\partial Z} = P - Y
$$

$$
\frac{\partial L}{\partial b} = \sum_{k=1}^{N} (P - Y)
$$

$$
\frac{\partial L}{\partial W} = X^T (P - Y)
$$

---

## 直感

- 勾配 = 誤差（p - y）
- Wの勾配 = 入力 × 誤差
- bの勾配 = 誤差そのもの

---

## ポイント

- 非常にシンプルな形になる
- バックプロパゲーションの基本構造
- すべての層で同じパターンが繰り返される

## パラメータ更新（勾配降下法）

$$
W := W - \eta \frac{\partial L}{\partial W}
$$

$$
b := b - \eta \frac{\partial L}{\partial b}
$$

ここで：

- $\eta$：学習率（learning rate）

##コード例

```python
import numpy as np

def softmax(x):
    x = x - np.max(x, axis=-1, keepdims=True)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)

def gradients(x, W, b, y_true):
    """
    x: (d,)
    W: (d, n)
    b: (n,)
    y_true: (n,) one-hot
    """
    z = x @ W + b
    p = softmax(z)

    dz = p - y_true  # (n,)

    dW = np.outer(x, dz)  # (d, n)
    db = dz

    return dW, db
```

```python
import numpy as np

def softmax(x):
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

# 初期化
x = np.array([1.0, 2.0])      # 入力
y = np.array([1, 0])          # 正解

W = np.random.randn(2, 2)
b = np.zeros(2)

lr = 0.1

# --- forward ---
z = x @ W + b
p = softmax(z)

# --- backward ---
dz = p - y
dW = np.outer(x, dz)
db = dz

# --- update ---
W -= lr * dW
b -= lr * db

print("p:", p)
print("updated W:\n", W)
```

# 2層ニューラルネットの勾配導出（直感＋数式）

## モデル構造

入力：
$$
x \in \mathbb{R}^d
$$

第1層（隠れ層）：
$$
z^{(1)} = x W^{(1)} + b^{(1)}
$$

活性化関数（ReLU）：
$$
h = \mathrm{ReLU}(z^{(1)})
$$

第2層（出力層）：
$$
z^{(2)} = h W^{(2)} + b^{(2)}
$$

softmax：
$$
p = \mathrm{softmax}(z^{(2)})
$$

損失関数：
$$
L = -\sum_i y_i \log p_i
$$

---

## 全体の流れ（図イメージ）

$$
x \rightarrow z^{(1)} \rightarrow h \rightarrow z^{(2)} \rightarrow p \rightarrow L
$$

---

## ステップ①：出力層の勾配

既知の結果：

$$
\frac{\partial L}{\partial z^{(2)}} = p - y
$$

---

## ステップ②：第2層の勾配

$$
\frac{\partial L}{\partial W^{(2)}} = h^T (p - y)
$$

$$
\frac{\partial L}{\partial b^{(2)}} = p - y
$$

---

## ステップ③：隠れ層への伝播

$$
\frac{\partial L}{\partial h} = (p - y) {W^{(2)}}^T
$$

---

## ステップ④：ReLUの勾配

ReLU：
$$
\mathrm{ReLU}(x) = \max(0, x)
$$

微分：

$$
\frac{d}{dx} \mathrm{ReLU}(x) =
\begin{cases}
1 & (x > 0) \\
0 & (x \leq 0)
\end{cases}
$$

---

$$
\frac{\partial L}{\partial z^{(1)}} =
\frac{\partial L}{\partial h} \odot \mathrm{ReLU}'(z^{(1)})
$$

---

## ステップ⑤：第1層の勾配

$$
\frac{\partial L}{\partial W^{(1)}} = x^T \frac{\partial L}{\partial z^{(1)}}
$$

$$
\frac{\partial L}{\partial b^{(1)}} = \frac{\partial L}{\partial z^{(1)}}
$$

---

## まとめ

$$
\frac{\partial L}{\partial z^{(2)}} = p - y
$$

$$
\frac{\partial L}{\partial W^{(2)}} = h^T (p - y)
$$

$$
\frac{\partial L}{\partial W^{(1)}} = x^T \left[ ((p - y) {W^{(2)}}^T) \odot \mathrm{ReLU}'(z^{(1)}) \right]
$$

---

## 直感

- 出力層：誤差 = p - y
- 隠れ層：誤差を逆伝播して調整
- 活性化関数：情報を通す or 止める

---

## 本質

勾配は後ろから前に流れる：

$$
L \leftarrow p \leftarrow z^{(2)} \leftarrow h \leftarrow z^{(1)} \leftarrow x
$$

---

## ポイント

- 誤差は後ろから伝わる
- 各層で「入力 × 誤差」の形になる
- 活性化関数が勾配の流れを制御する

##コード例

```python
import numpy as np

def relu(x):
    return np.maximum(0, x)

def relu_grad(x):
    return (x > 0).astype(float)

def softmax(x):
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

def forward(x, W1, b1, W2, b2):
    z1 = x @ W1 + b1
    h = relu(z1)
    z2 = h @ W2 + b2
    p = softmax(z2)
    return z1, h, z2, p

def backward(x, y, z1, h, z2, p, W2):
    dz2 = p - y

    dW2 = np.outer(h, dz2)
    db2 = dz2

    dh = dz2 @ W2.T
    dz1 = dh * relu_grad(z1)

    dW1 = np.outer(x, dz1)
    db1 = dz1

    return dW1, db1, dW2, db2
```

# 多層ニューラルネット（3層以上）の一般化

## モデル（L層）

入力：
$$
a^{(0)} = x
$$

各層（l = 1, ..., L）：

$$
z^{(l)} = a^{(l-1)} W^{(l)} + b^{(l)}
$$

$$
a^{(l)} = f^{(l)}(z^{(l)})
$$

出力層（softmax）：
$$
a^{(L)} = p
$$

---

## 全体の流れ

$$
x \rightarrow z^{(1)} \rightarrow a^{(1)} \rightarrow z^{(2)} \rightarrow \cdots \rightarrow z^{(L)} \rightarrow p \rightarrow L
$$

---

## ステップ①：出力層

$$
\delta^{(L)} = \frac{\partial L}{\partial z^{(L)}} = p - y
$$

---

## ステップ②：逆伝播（一般形）

各層について：

$$
\delta^{(l)} =
\left( \delta^{(l+1)} {W^{(l+1)}}^T \right)
\odot f'(z^{(l)})
$$

---

## ステップ③：勾配

$$
\frac{\partial L}{\partial W^{(l)}} =
{a^{(l-1)}}^T \delta^{(l)}
$$

$$
\frac{\partial L}{\partial b^{(l)}} =
\delta^{(l)}
$$

---

## まとめ（最重要）

$$
\delta^{(L)} = p - y
$$

$$
\delta^{(l)} = (\delta^{(l+1)} W^{(l+1)T}) \odot f'(z^{(l)})
$$

$$
\frac{\partial L}{\partial W^{(l)}} = a^{(l-1)T} \delta^{(l)}
$$

---

## 直感

- 誤差は後ろから前に伝わる
- 各層で「入力 × 誤差」で重み更新
- 活性化関数が勾配の流れを制御

---

## 本質

バックプロパゲーションとは：

$$
誤差をチェーンルールで逆方向に伝えるだけ
$$

---

## 図イメージ

$$
L \leftarrow a^{(L)} \leftarrow z^{(L)} \leftarrow \cdots \leftarrow z^{(1)} \leftarrow x
$$

---

## ポイント

- どれだけ層が増えても同じ式
- 「δ（デルタ）」を伝播するだけ
- 計算はすべて行列演算で書ける

##コード例

```python
import numpy as np

def relu(x):
    return np.maximum(0, x)

def relu_grad(x):
    return (x > 0).astype(float)

def softmax(x):
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

# forward
def forward(x, weights, biases):
    """
    weights: list of W
    biases: list of b
    """
    activations = [x]
    zs = []

    a = x
    for i in range(len(weights) - 1):
        z = a @ weights[i] + biases[i]
        zs.append(z)
        a = relu(z)
        activations.append(a)

    # 最後はsoftmax
    z = a @ weights[-1] + biases[-1]
    zs.append(z)
    a = softmax(z)
    activations.append(a)

    return activations, zs


# backward
def backward(x, y, weights, biases, activations, zs):
    L = len(weights)

    dWs = [None] * L
    dbs = [None] * L

    # 出力層
    delta = activations[-1] - y

    for l in reversed(range(L)):
        a_prev = activations[l]

        dWs[l] = np.outer(a_prev, delta)
        dbs[l] = delta

        if l > 0:
            delta = (delta @ weights[l].T) * relu_grad(zs[l-1])

    return dWs, dbs
```

# 勾配消失問題（Vanishing Gradient）

## 現象

ネットワークが深くなると、前の層に勾配がほとんど届かなくなる：

$$
\frac{\partial L}{\partial W^{(1)}} \approx 0
$$

→ 重みが更新されない → 学習が進まない

---

## 原因（チェーンルールの連続）

バックプロパゲーションでは：

$$
\delta^{(l)} =
\left( \delta^{(l+1)} W^{(l+1)T} \right)
\odot f'(z^{(l)})
$$

これを展開すると：

$$
\delta^{(1)} =
\delta^{(L)} \cdot
\prod_{k=2}^{L} \left( W^{(k)} f'(z^{(k-1)}) \right)
$$

---

## 何が起きているか

- 多数の「小さい値」を掛け続ける
- 結果として指数的に小さくなる

$$
0.5 \times 0.5 \times 0.5 \times \cdots \rightarrow 0
$$

---

## sigmoid の問題

sigmoid：

$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$

導関数：

$$
\sigma'(x) = \sigma(x)(1 - \sigma(x)) \le 0.25
$$

---

## 結論

各層で最大でも 0.25 倍：

$$
(0.25)^L \rightarrow 0
$$

層が深くなると：

$$
\delta \rightarrow 0
$$

---

## 直感

- 誤差が後ろから伝わる
- 途中でどんどん弱くなる
- 最初の層まで届かない

---

## 何が問題か

- 初期層が学習されない
- 特徴抽出ができない
- モデルが賢くならない

---

## 解決策

### 1. ReLU

$$
f'(x) =
\begin{cases}
1 & (x > 0) \\
0 & (x \le 0)
\end{cases}
$$

→ 勾配が減衰しにくい

## 本質

勾配消失とは：

$$
小さい値の連続積による指数的減衰
$$

##コード例

```python
import numpy as np

# sigmoidの微分
def sigmoid_grad(x):
    s = 1 / (1 + np.exp(-x))
    return s * (1 - s)

# 勾配の減衰をシミュレーション
layers = 20
grad = 1.0

print("layer  grad")

for i in range(layers):
    g = sigmoid_grad(np.random.randn())
    grad *= g
    print(f"{i+1:2d}    {grad:.10f}")
```

```python
def relu_grad(x):
    return np.where(x > 0, 1.0, 0.0)

grad = 1.0

for i in range(20):
    g = relu_grad(np.random.randn())
    grad *= g
    print(grad)
```
