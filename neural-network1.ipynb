#勾配を求める

```python
def numerical_gradient(f, x):
  """
  Parameters
  ----------
  f : function
      入力: numpy配列
      出力: numpy配列
  x : numpy.ndarray
  """
  h = 1e-4
  grad = np.zeros_like(x)

  for idx in range(x.size):
    tmp_val = x[idx]
    # f(x+h)の計算
    x[idx] = tmp_val + h
    fxh1 = f(x)

    # f(x-h)の計算
    x[idx] = tmp_val - h
    fxh2 = f(x)

    grad[idx] = (fxh1 - fxh2) / (2*h)
    x[idx] = tmp_val

  return grad
```

```python
def func(x):
  """
  2次元配列(numpy配列)
  """
  return x[0]**2 + x[1]**2

print("例1:", numerical_gradient(func, np.array([3.0, 4.0])))
print("例2:", numerical_gradient(func, np.array([0.0, 2.0])))
print("例3:", numerical_gradient(func, np.array([3.0, 0.0])))
```

#勾配降下法

```python
def gradient_descent(f, init_x, lr=0.01, step_num=100):
  """
  f: function
  init_x: xの初期値
  lr: 学習率
  step_num: 学習回数
  """
  x = init_x

  for i in range(step_num):
    grad = numerical_gradient(f, x)
    x -= lr * grad

  return x
```

##例）関数の極小値を求める

```python
init_x = np.array([-3.0, 4.0])
x = gradient_descent(func, init_x, lr=0.1, step_num=100)

print(x)
```

```python
func(x)
```

funcの極小値は0なので、ほぼ正解に近い値を求められている

ハイパーパラメータの学習率lrを小さい値に設定すると上手くいかない

```python
init_x = np.array([-3.0, 4.0])
x = gradient_descent(func, init_x, lr=0.001, step_num=100)

print("x:", x)
print("func:", func(x))
```

#ソフトマックス関数・クロスエントロピー関数

## ソフトマックス関数

$$
\mathrm{softmax}(x_i) = \frac{e^{x_i}}{\sum_{j=1}^{n} e^{x_j}}
$$

数値安定化のための実装では、以下を用いる：

$$
\mathrm{softmax}(x_i) = \frac{e^{x_i - \max(x)}}{\sum_{j=1}^{n} e^{x_j - \max(x)}}
$$



```python
import numpy as np

def softmax(x):
    """
    x: numpy.ndarray (n,) または (batch, n)
    return: softmax適用結果（同じshape）
    """
    # 数値安定化（オーバーフロー防止）
    x = x - np.max(x, axis=-1, keepdims=True)

    exp_x = np.exp(x)
    sum_exp = np.sum(exp_x, axis=-1, keepdims=True)

    return exp_x / sum_exp
```

## クロスエントロピー損失関数

分類問題におけるクロスエントロピー：

$$
L = - \sum_{i=1}^{n} y_i \log p_i
$$

ここで：

- $y_i$：正解ラベル（one-hotベクトル）
- $p_i$：予測確率（softmaxの出力）

---

バッチデータの場合：

$$
L = - \frac{1}{N} \sum_{k=1}^{N} \sum_{i=1}^{n} y_{k,i} \log p_{k,i}
$$

```python
import numpy as np

def cross_entropy(y_true, y_pred):
    """
    y_true: (n,) または (batch, n) one-hot
    y_pred: (n,) または (batch, n) softmax出力

    return: スカラー損失
    """
    # log(0)防止
    eps = 1e-15
    y_pred = np.clip(y_pred, eps, 1 - eps)

    return -np.sum(y_true * np.log(y_pred)) / y_true.shape[0]
```
